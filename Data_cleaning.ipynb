```python
## 3. Missing Data Handling

### Age

The `Age` column contains missing values. Median imputation will be used because age is a numerical variable and the median is less affected by extreme values than the mean. This helps preserve the available employee records without allowing potential outliers to significantly influence the imputed values.

### Salary

The `Salary` column also contains missing values. Median imputation will be used because salary data can be skewed by employees with unusually high salaries. The median therefore provides a more robust estimate of a typical salary.

### Row Deletion

Rows will not be deleted solely because `Age` or `Salary` is missing. Since only a portion of the dataset contains missing values, imputation allows us to retain useful employee records.
```

```python
# Store median values before imputation

age_median = df["Age"].median()
salary_median = df["Salary"].median()

print("Age median:", age_median)
print("Salary median:", salary_median)
```

```python
# Fill missing numerical values with median

df["Age"] = df["Age"].fillna(age_median)
df["Salary"] = df["Salary"].fillna(salary_median)
```

```python
## 5. Outlier Detection

The Interquartile Range (IQR) method is used to identify potential outliers in numerical columns.

The lower and upper boundaries are calculated using the first quartile (Q1), third quartile (Q3), and IQR. Observations outside these boundaries are considered potential outliers.

Outliers will not automatically be deleted because an extreme value may represent a genuine employee record. The detected observations will therefore be reviewed before deciding whether to retain, cap, or remove them.
```

```python
numeric_columns = ["Age", "Salary"]

outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]
    
    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

salary_outliers = df[
    (df["Salary"] < lower) |
    (df["Salary"] > upper)
]

salary_outliers[["Employee_ID", "Salary"]]
```
